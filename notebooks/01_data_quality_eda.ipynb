# Calidad y análisis exploratorio del dataset YouToxic

## Objetivo

Evaluar la calidad del dataset antes de su limpieza, división y procesamiento para el modelo `IsToxic`.

El análisis comprobará:

- estructura y tipos de datos;
- valores ausentes;
- identificadores y duplicados;
- distribución de `IsToxic`;
- longitud de los textos;
- diferencias entre vídeos;
- riesgo de fuga de información;
- viabilidad de una división agrupada por `VideoId`.

## Privacidad y alcance

- El CSV permanece local y no se versiona.
- El notebook no mostrará comentarios reales.
- Solo se guardarán estadísticas agregadas.
- Toxicidad no se interpretará como sentimiento, intención ni infracción de políticas.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

RANDOM_STATE = 42

PROJECT_ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

DATASET_PATH = Path(
    os.getenv(
        "DATASET_PATH",
        PROJECT_ROOT / "data" / "raw" / "youtoxic_english_1000.csv",
    )
).resolve()

pd.set_option("display.max_columns", 20)
pd.set_option("display.max_rows", 30)

sns.set_theme(style="whitegrid", context="notebook")

if not DATASET_PATH.is_file():
    raise FileNotFoundError(f"Dataset no encontrado: {DATASET_PATH}")

print(f"Python y librerías cargadas correctamente")
print(f"Dataset encontrado: {DATASET_PATH.name}")
```

```python
dataset = pd.read_csv(DATASET_PATH)

required_columns = {"CommentId", "VideoId", "Text", "IsToxic"}
missing_required_columns = required_columns - set(dataset.columns)

if missing_required_columns:
    raise ValueError(
        f"Faltan columnas obligatorias: {sorted(missing_required_columns)}"
    )

schema_summary = pd.DataFrame(
    {
        "dtype": dataset.dtypes.astype(str),
        "missing_values": dataset.isna().sum(),
        "missing_percentage": (
            dataset.isna().mean().mul(100).round(2)
        ),
        "unique_values": dataset.nunique(dropna=False),
    }
)

print(f"Filas: {len(dataset):,}")
print(f"Columnas: {dataset.shape[1]}")
print(f"Columnas obligatorias presentes: {not missing_required_columns}")

display(schema_summary)
```
